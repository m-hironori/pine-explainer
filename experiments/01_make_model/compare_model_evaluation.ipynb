# Compare model evaluation

```python
matcher_names = ["bert_mini", "magellan"]
dataset_root_dir = "../../data/lemon/datasets"
model_root_dir = "../../data/lemon/model"
dataset_names = [
    "structured_amazon_google",
    "structured_beer",
    "structured_dblp_acm",
    "structured_dblp_google_scholar",
    "structured_fodors_zagat",
    "structured_walmart_amazon",
    "structured_itunes_amazon",
    "dirty_dblp_acm",
    "dirty_dblp_google_scholar",
    "dirty_walmart_amazon",
    "dirty_itunes_amazon",
    "textual_abt_buy",
    "textual_company",
]
```

```python
import pathlib
import pickle
from typing import Dict

def load_model_evaluation(dataset_name:str, matcher_name:str, model_root_dir:str)->Dict:
    if matcher_name == "bert_mini":
        matcher_name = "bert-mini"
    model_eval_path = pathlib.Path(model_root_dir) / matcher_name / dataset_name / "eval_result.pickle"
    # {'precision': xx , 'recall': xx, 'f1': xx}
    with model_eval_path.open("rb") as f:
        eval_results = pickle.load(f)
    # PrecとRecallが逆なので入れ替え
    tmp = eval_results["precision"]
    eval_results["precision"] = eval_results["recall"]
    eval_results["recall"] = tmp
    return eval_results


def test_load_model_evaluation():
    dataset_name = dataset_names[0]
    matcher_name = matcher_names[0]
    ret = load_model_evaluation(dataset_name, matcher_name, model_root_dir)
    assert ret.get("precision", None) is not None
    assert ret.get("recall", None) is not None
    assert ret.get("f1", None) is not None

    dataset_name = dataset_names[1]
    matcher_name = matcher_names[1]
    ret = load_model_evaluation(dataset_name, matcher_name, model_root_dir)
    assert ret.get("precision", None) is not None
    assert ret.get("recall", None) is not None
    assert ret.get("f1", None) is not None

test_load_model_evaluation()
```

```python
import pandas as pd

df_results_dict = {}
for matcher_name in matcher_names:
    results = []
    for dataset_name in dataset_names:
        result = load_model_evaluation(dataset_name, matcher_name, model_root_dir)
        results.append(result)
    df_result = pd.DataFrame.from_records(results, index=dataset_names)
    df_results_dict[matcher_name] = df_result

display(pd.concat(df_results_dict.values(), axis=1, keys=df_results_dict.keys()))
```
